## Simple techniques for Data preprocessing 

```python
import numpy as np 
import pandas as pd

```

### Generating random data

```python
np.random.seed(42)
def createdata():
    data ={
        "Age":np.random.randint(18,70,size=20),
        "Salary":np.random.randint(30000,120000,size=20),
        "Purchased":np.random.choice([0,1],size=20),
        "Gender":np.random.choice(["male","female"],size=20),
        "City":np.random.choice(["New York","San Francisco","Los Angeles"],size=20)
        }
    df = pd.DataFrame(data)
    return df
df= createdata()
df.head(10)
```

```python
df.shape
```

### when we have some missing values

```python
# Introduce some missing values 
df.loc[4,"Age"] = np.nan
df.loc[10,"Salary"]= np.nan
df.head(11)
```

```python
#  simple Imputaion(mean, median, mode)
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
df.head(11)
```

```python
df_dropped = df.dropna()
df_dropped.head()
```

```python
df_dropped.shape
```

```python
from sklearn.impute import KNNImputer
knn_imputer = KNNImputer(n_neighbors=3)
df[["Age","Salary"]]=knn_imputer.fit_transform(df[["Age","Salary"]])
```
